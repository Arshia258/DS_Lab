## 3. The K-Means algorithm and its objective function

### 3.1 Steps

1. **Choose K** – the number of clusters.
2. **Initialise** K centroids $\mu_1, \dots, \mu_K$ (randomly picked points, or the smarter *k-means++* in scikit-learn).
3. **Assignment step** – give each point $x_i$ to the closest centroid using **Euclidean distance**:
$$d(x, \mu) = \sqrt{(x_1-\mu_1)^2 + (x_2-\mu_2)^2 + \dots}$$
$$c_i = \arg\min_{k}\; d(x_i, \mu_k)$$
4. **Update step** – move every centroid to the **mean** of the points assigned to it:
$$\mu_k = \frac{1}{|C_k|}\sum_{x_i \in C_k} x_i$$
5. **Repeat** steps 3–4 until the assignments (or centroids) stop changing — this is **convergence**.

### 3.2 What K-Means is minimising

K-Means minimises the **Within-Cluster Sum of Squares (WCSS)**, which scikit-learn calls **`inertia_`**:

$$\text{WCSS} = \sum_{k=1}^{K}\;\sum_{x_i \in C_k} \lVert x_i - \mu_k \rVert^2$$

* Each assignment step and each update step can only **decrease (or keep)** the WCSS ⇒ the algorithm is guaranteed to converge.
* But it converges to a **local** minimum, which depends on the starting centroids. That is why scikit-learn runs it several times (`n_init`) and keeps the best result.

## 4. Numerical worked example (by hand) ✍️

Let's cluster **7 points** into **K = 2** clusters completely by hand, and then let Python confirm every number.

| Point | x | y |
|---|---|---|
| A | 1.0 | 1.0 |
| B | 1.5 | 2.0 |
| C | 3.0 | 4.0 |
| D | 5.0 | 7.0 |
| E | 3.5 | 5.0 |
| F | 4.5 | 5.0 |
| G | 3.5 | 4.5 |

**Initial centroids** (chosen as two far-apart points): $\mu_1 = A = (1, 1)$ and $\mu_2 = D = (5, 7)$.

---

### 🔁 Iteration 1

**Assignment step.** Example calculation for point **B (1.5, 2)**:

$$d(B,\mu_1) = \sqrt{(1.5-1)^2 + (2-1)^2} = \sqrt{0.25+1} = \sqrt{1.25} = 1.118$$
$$d(B,\mu_2) = \sqrt{(1.5-5)^2 + (2-7)^2} = \sqrt{12.25+25} = \sqrt{37.25} = 6.103$$

$1.118 < 6.103$ ⇒ **B → Cluster 1**. Doing this for every point:

| Point | d to μ₁ (1, 1) | d to μ₂ (5, 7) | Assigned to |
|---|---|---|---|
| A | 0.000 | 7.211 | **C1** |
| B | 1.118 | 6.103 | **C1** |
| C | 3.606 | 3.606 | **C1** *(tie → first centroid)* |
| D | 7.211 | 0.000 | **C2** |
| E | 4.717 | 2.500 | **C2** |
| F | 5.315 | 2.062 | **C2** |
| G | 4.301 | 2.915 | **C2** |

> ⚠️ Point **C** is *exactly* the same distance from both centroids. By convention (and in NumPy's `argmin`) ties go to the **first** centroid.

**Update step.**

$$\mu_1 = \text{mean}(A,B,C) = \left(\tfrac{1+1.5+3}{3},\; \tfrac{1+2+4}{3}\right) = (1.833,\; 2.333)$$
$$\mu_2 = \text{mean}(D,E,F,G) = \left(\tfrac{5+3.5+4.5+3.5}{4},\; \tfrac{7+5+5+4.5}{4}\right) = (4.125,\; 5.375)$$

---

### 🔁 Iteration 2

| Point | d to μ₁ (1.833, 2.333) | d to μ₂ (4.125, 5.375) | Assigned to |
|---|---|---|---|
| A | 1.572 | 5.376 | C1 |
| B | 0.471 | 4.276 | C1 |
| C | 2.034 | **1.777** | **C2** ⬅ *changed!* |
| D | 5.640 | 1.846 | C2 |
| E | 3.145 | 0.729 | C2 |
| F | 3.771 | 0.530 | C2 |
| G | 2.734 | 1.075 | C2 |

Point **C** has moved to Cluster 2. **Update step:**

$$\mu_1 = \text{mean}(A,B) = (1.25,\; 1.5)$$
$$\mu_2 = \text{mean}(C,D,E,F,G) = \left(\tfrac{3+5+3.5+4.5+3.5}{5},\; \tfrac{4+7+5+5+4.5}{5}\right) = (3.9,\; 5.1)$$

---

### 🔁 Iteration 3

| Point | d to μ₁ (1.25, 1.5) | d to μ₂ (3.9, 5.1) | Assigned to |
|---|---|---|---|
| A | 0.559 | 5.022 | C1 |
| B | 0.559 | 3.920 | C1 |
| C | 3.052 | 1.421 | C2 |
| D | 6.657 | 2.195 | C2 |
| E | 4.161 | 0.412 | C2 |
| F | 4.776 | 0.608 | C2 |
| G | 3.750 | 0.721 | C2 |

No point changed cluster ⇒ the centroids stay at $(1.25, 1.5)$ and $(3.9, 5.1)$ ⇒ **converged ✅**

**Final clusters:** C1 = {A, B}, C2 = {C, D, E, F, G}

### 📉 WCSS at the end

$$\text{WCSS} = \underbrace{(0.25^2+0.5^2) + (0.25^2+0.5^2)}_{\text{C1: } A, B = 0.625} + \underbrace{(0.9^2+1.1^2) + (1.1^2+1.9^2) + (0.4^2+0.1^2) + (0.6^2+0.1^2) + (0.4^2+0.6^2)}_{\text{C2} = 7.9} = 8.525$$

WCSS went **33.25 → 11.23 → 8.525** across the iterations (computed with the centroids used in each assignment step) — it never increases.

## 5. Evaluation metrics for clustering 📏

Because clustering has **no ground-truth labels**, we can't use accuracy. Instead we use:

* **Internal metrics** – judge the clusters using only the data (compactness & separation).
* **External metrics** – compare to true labels, *only if* they happen to be available (e.g. for teaching datasets like Iris).

### 5.1 Internal metrics

| Metric | Formula / idea | Range | Better when | In sklearn |
|---|---|---|---|---|
| **WCSS / Inertia** | $\sum_k \sum_{x \in C_k}\lVert x-\mu_k\rVert^2$ — compactness | $[0, \infty)$ | **Lower** (but *always* falls as K grows) | `km.inertia_` |
| **Silhouette score** | $s(i) = \dfrac{b(i) - a(i)}{\max\{a(i), b(i)\}}$ averaged over all points | $[-1, 1]$ | **Higher** (≈1 great, ≈0 overlapping, <0 wrong cluster) | `silhouette_score` |


In the silhouette:
* $a(i)$ = average distance from point $i$ to the **other points in its own cluster** (how tight its cluster is),
* $b(i)$ = average distance from point $i$ to the points of the **nearest other cluster** (how far the neighbours are).

### 5.2 External metrics (need true labels)

| Metric | Range | Meaning |
|---|---|---|
| **Adjusted Rand Index (ARI)** | $[-1, 1]$ | 1 = perfect match with true labels, ≈0 = random labelling |
| **Normalized Mutual Information (NMI)** | $[0, 1]$ | 1 = clusters carry exactly the same information as the labels |

Both ignore the *names* of the clusters — cluster "0" vs "1" swapped still scores 1.0.

---

### ✍️ Hand calculation: silhouette of point A in the worked example

Final clusters: C1 = {A, B}, C2 = {C, D, E, F, G}.

**a(A)** – A's own cluster contains only B:
$$a(A) = d(A,B) = 1.118$$

**b(A)** – average distance from A to every point of C2:
$$b(A) = \frac{d(A,C)+d(A,D)+d(A,E)+d(A,F)+d(A,G)}{5} = \frac{3.606+7.211+4.717+5.315+4.301}{5} = \frac{25.150}{5} = 5.030$$

$$s(A) = \frac{5.030 - 1.118}{\max(1.118,\; 5.030)} = \frac{3.912}{5.030} = \mathbf{0.778}$$

A is **well inside** its cluster (close to 1). Let's verify, and compute all metrics for the worked example:

# Z-score normalization

````python
# Z-score normalization
X_scaled = (X - X.mean(axis=0)) / X.std(axis=0)

print("Standardized Data:\n", X_scaled[:5])
````

# Define the custom K-Means algorithm

````python
def kmeans(X, k, max_iters=100):
    np.random.seed(42)

    centroids = X[np.random.choice(len(X), k, replace=False)].copy()

    for _ in range(max_iters):
        labels = np.argmin(
            np.array([np.sum((X-c)**2, axis=1) for c in centroids]).T,
            axis=1
        )

        new_centroids = np.array([
            X[labels == i].mean(axis=0) if np.any(labels == i)
            else centroids[i] for i in range(k)
        ])

        if np.allclose(centroids, new_centroids):
            break

        centroids = new_centroids

    labels = np.argmin(
        np.array([np.sum((X-c)**2, axis=1) for c in centroids]).T,
        axis=1
    )

    return centroids, labels
````

````python
k = 3
centroids, labels = kmeans(X_scaled, k)

df['Cluster'] = labels

print(df.head())
print("Centroids:\n", centroids)
````

````python
# Complete Wine dataset lab assignment
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans as SKKMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score, normalized_mutual_info_score
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Load data. Keep true labels aside; do not use them for clustering.
wine = load_wine()
X_wine = wine.data
y_true = wine.target
wine_df = pd.DataFrame(X_wine, columns=wine.feature_names)
print("Shape:", wine_df.shape)
display(wine_df.head())
display(wine_df.describe())

# Step 2: Standardise all 13 features.
# Scaling is necessary because features have very different units and ranges.
# Without scaling, large-range features (such as proline) dominate Euclidean distance.
scaler_wine = StandardScaler()
X_wine_scaled = scaler_wine.fit_transform(X_wine)

# PCA is used only to visualise the 13-dimensional data in 2 dimensions.
pca = PCA(n_components=2, random_state=42)
X_wine_2d = pca.fit_transform(X_wine_scaled)
print("Explained variance ratio (2 PCs):", pca.explained_variance_ratio_)
print("Total variance explained:", pca.explained_variance_ratio_.sum())

# WCSS helper function
def calculate_wcss(X, labels, centroids):
    return float(sum(((X[labels == j] - centroids[j]) ** 2).sum()
                     for j in range(len(centroids))))

# Step 3: NumPy-only K-Means from scratch
def kmeans_scratch(X, k, max_iters=100, n_init=10, seed=42):
    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or not 1 <= k <= len(X):
        raise ValueError("X must be 2D and k must be between 1 and n_samples.")
    rng = np.random.default_rng(seed)
    best = None

    for run in range(n_init):
        centroids = X[rng.choice(len(X), size=k, replace=False)].copy()
        iterations = 0

        for iteration in range(max_iters):
            distances = ((X[:, None, :] - centroids[None, :, :]) ** 2).sum(axis=2)
            labels = np.argmin(distances, axis=1)
            new_centroids = centroids.copy()

            for j in range(k):
                members = X[labels == j]
                if len(members):
                    new_centroids[j] = members.mean(axis=0)
                else:
                    new_centroids[j] = X[rng.integers(len(X))]

            iterations = iteration + 1
            if np.allclose(new_centroids, centroids):
                centroids = new_centroids
                break
            centroids = new_centroids

        distances = ((X[:, None, :] - centroids[None, :, :]) ** 2).sum(axis=2)
        labels = np.argmin(distances, axis=1)
        wcss = calculate_wcss(X, labels, centroids)

        if best is None or wcss < best[2]:
            best = (centroids.copy(), labels.copy(), wcss, iterations)

    return best

# Bonus: calculate the mean silhouette score using NumPy only.
def silhouette_scratch(X, labels):
    X = np.asarray(X, dtype=float)
    labels = np.asarray(labels)
    unique_labels = np.unique(labels)
    if len(unique_labels) < 2 or len(unique_labels) >= len(X):
        raise ValueError("Silhouette requires 2 to n_samples-1 clusters.")

    # Pairwise Euclidean distance matrix
    distances = np.sqrt(((X[:, None, :] - X[None, :, :]) ** 2).sum(axis=2))
    scores = []

    for i in range(len(X)):
        own = labels == labels[i]
        own[i] = False
        a = distances[i, own].mean() if own.any() else 0.0

        other_means = [
            distances[i, labels == label].mean()
            for label in unique_labels if label != labels[i]
        ]
        b = min(other_means)
        denom = max(a, b)
        scores.append((b - a) / denom if denom > 0 else 0.0)

    return float(np.mean(scores))

# Step 4: Compare scratch and scikit-learn for K=3.
k = 3
cent_scratch, labels_scratch, wcss_scratch, iters_scratch = kmeans_scratch(
    X_wine_scaled, k, max_iters=100, n_init=10, seed=42
)
km_wine = SKKMeans(n_clusters=k, n_init=10, random_state=42)
labels_sk = km_wine.fit_predict(X_wine_scaled)

sil_scratch = silhouette_scratch(X_wine_scaled, labels_scratch)
sil_scratch_sk = silhouette_score(X_wine_scaled, labels_scratch)
sil_sk = silhouette_score(X_wine_scaled, labels_sk)

print("Scratch centroids:\\n", cent_scratch)
print("Scratch WCSS:", wcss_scratch, "| iterations:", iters_scratch)
print("scikit-learn centroids:\\n", km_wine.cluster_centers_)
print("scikit-learn inertia:", km_wine.inertia_, "| iterations:", km_wine.n_iter_)
print("Scratch silhouette (NumPy):", sil_scratch)
print("Scratch silhouette (sklearn check):", sil_scratch_sk)
print("scikit-learn silhouette:", sil_sk)
print("External metrics (evaluation only; labels were not used to fit):")
print("Scratch ARI:", adjusted_rand_score(y_true, labels_scratch))
print("Scratch NMI:", normalized_mutual_info_score(y_true, labels_scratch))
print("scikit-learn ARI:", adjusted_rand_score(y_true, labels_sk))
print("scikit-learn NMI:", normalized_mutual_info_score(y_true, labels_sk))

# Step 5: Elbow and silhouette curves for K=2,...,10.
k_values = range(2, 11)
scratch_wcss, sklearn_wcss = [], []
scratch_sil, sklearn_sil = [], []

for current_k in k_values:
    c, lab, wc, it = kmeans_scratch(
        X_wine_scaled, current_k, max_iters=100, n_init=10,
        seed=42 + current_k
    )
    scratch_wcss.append(wc)
    scratch_sil.append(silhouette_score(X_wine_scaled, lab))

    model = SKKMeans(n_clusters=current_k, n_init=10, random_state=42)
    lab_sk = model.fit_predict(X_wine_scaled)
    sklearn_wcss.append(model.inertia_)
    sklearn_sil.append(silhouette_score(X_wine_scaled, lab_sk))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(list(k_values), scratch_wcss, 'o-', label='From scratch')
axes[0].plot(list(k_values), sklearn_wcss, 's--', label='scikit-learn')
axes[0].set(title='Elbow Method', xlabel='K', ylabel='WCSS')
axes[0].legend()
axes[0].grid(alpha=0.25)
axes[1].plot(list(k_values), scratch_sil, 'o-', label='From scratch')
axes[1].plot(list(k_values), sklearn_sil, 's--', label='scikit-learn')
axes[1].set(title='Silhouette Scores', xlabel='K', ylabel='Silhouette score')
axes[1].legend()
axes[1].grid(alpha=0.25)
plt.tight_layout()
plt.show()

# PCA visualisation of scikit-learn clusters (clustering was done in all 13 features).
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_wine_2d[:, 0], X_wine_2d[:, 1],
                      c=labels_sk, cmap='viridis', edgecolor='k', alpha=0.8)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Wine dataset: K-Means clusters visualised using PCA')
plt.colorbar(scatter, label='Cluster')
plt.show()

# Step 6: Metric comparison table for chosen K=3.
comparison = pd.DataFrame({
    'Metric': ['WCSS / Inertia', 'Silhouette score'],
    'Type': ['Internal', 'Internal'],
    'Better': ['Lower', 'Higher (maximum 1)'],
    'Scratch': [wcss_scratch, sil_scratch],
    'scikit-learn': [km_wine.inertia_, sil_sk]
})
display(comparison)

# State the choice based on the curves. K=3 is also the known number of
# cultivars, but the true labels were not used to train either model.
print("Chosen K for this lab: 3. Compare the elbow and silhouette curves above;")
print("K=3 corresponds to the three known wine cultivars, assessed here only")
print("after clustering through the external ARI/NMI metrics.")
````
